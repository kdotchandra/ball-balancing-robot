```python
import numpy as np
from control import lqr, ss

# พารามิเตอร์ตัวอย่าง
m = 0.05      # mass ball
M = 1.0       # effective mass of platform
g = 9.81
k_c = 2000.0  # contact stiffness (N/m) - ปรับตามจริง
c_c = 50.0    # contact damping

# สถานะ x = [z_r, z_r_dot, h, h_dot]
A = np.array([
  [0, 1, 0, 0],
  [-k_c/m, -c_c/m, k_c/m, c_c/m],
  [0, 0, 0, 1],
  [k_c/M, c_c/M, -k_c/M, -c_c/M]
])
B = np.array([[0], [0], [0], [1.0/M]])  # actuator applies force to the platform

# เลือก Q,R
Q = np.diag([1000, 100, 10, 1])   # penalize z_r and z_r_dot heavily
R = np.array([[0.01]])           # penalize large control effort

K, S, E = lqr(A, B, Q, R)
print("K =", K)
print("Closed-loop poles =", E)
```
